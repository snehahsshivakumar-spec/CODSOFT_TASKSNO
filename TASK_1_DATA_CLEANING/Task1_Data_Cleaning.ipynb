## Data Cleaning and Preprocessing

### Dataset: Retail Sales Data

This project focuses on cleaning and preprocessing a retail sales dataset using Python and Pandas.

### Objectives
- Import and inspect the dataset
- Identify missing values
- Identify duplicate records
- Detect inconsistent data entries
- Correct data types
- Clean and preprocess the dataset
- Save the cleaned dataset for further analysis

### Tools Used
- Python
- Pandas
- NumPy
- Jupyter Notebook

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("CODSOFT_Task1_Retail_Sales_Raw.csv")
df.head()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df["Quantity"].median()
```

```python
df["Customer_Name"] = df["Customer_Name"].fillna("Unknown")

df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
df.isnull().sum()
```

```python
df["Category"].unique()
```

```python
df["Payment_Method"].unique()
```

```python
df["Category"] = df["Category"].replace({
    "home": "Home",
    "ELECTRONICS": "Electronics"
})
```

```python
df["Payment_Method"] = df["Payment_Method"].replace({
    "upi": "UPI"
})
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
df["Date"].dtype
```

```python
df["Unit_Price"].unique()
```

```python
df["Unit_Price"] = df["Unit_Price"].astype(str).str.replace(" INR", "", regex=False).str.strip()

df["Unit_Price"] = pd.to_numeric(df["Unit_Price"], errors="coerce")
```
